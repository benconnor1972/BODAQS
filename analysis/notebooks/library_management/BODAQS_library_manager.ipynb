# BODAQS Library Manager

Notebook UI for inspecting a processed library, editing session notes, and creating or updating root-level Study Sets. Legacy aggregations are no longer the forward path; use Study Set groupings instead.

## 1. Configure Library

Set `LIBRARIES_ROOT` to the directory that contains one or more BODAQS libraries, then set `LIBRARY_ID` to the library you want to manage.

```python
from pathlib import Path
import sys
from IPython.display import display


def _find_analysis_root(start: Path) -> Path:
    for candidate in (start, *start.parents):
        if (candidate / "bodaqs_analysis").is_dir():
            return candidate
        analysis = candidate / "analysis"
        if (analysis / "bodaqs_analysis").is_dir():
            return analysis
    raise RuntimeError("Could not locate the local bodaqs_analysis package")


ANALYSIS_ROOT = _find_analysis_root(Path.cwd().resolve())
if str(ANALYSIS_ROOT) not in sys.path:
    sys.path.insert(0, str(ANALYSIS_ROOT))

from bodaqs_analysis.ui import make_library_manager

LIBRARIES_ROOT = Path(r"C:\Users\benco\OneDrive\BODAQS-data")
LIBRARY_ID = "archie"
```

## 2. Open Manager

The `Sessions` tab uses the Library API catalog and note API. The `Study Sets` tab writes root-level Study Set JSON files compatible with the web app.

```python
mgr = make_library_manager(
    libraries_root=LIBRARIES_ROOT,
    library_id=LIBRARY_ID,
)
display(mgr["ui"])
```
