# BODAQS Simple Suspension Metrics

This notebook generates displacement and velocity metrics and stats from processed log data. Select the session or sessions you want to view: if you choose more than one, you can see a comparison between the selected sessions. 

Use the first code cell to choose the artifacts directory, then run the second code cell to load the selector and dashboard.

If you are starting with raw logger data, then either run the batch pre-processor to add it to the library or use the `bodaqs_one_step_suspension_metrics` notebook.  

```python
from pathlib import Path
from IPython.display import display
from bodaqs_analysis.dashboards import make_simple_suspension_metrics_dashboard
from bodaqs_analysis.ui import make_preprocess_runtime_settings_editor
from bodaqs_analysis.widgets.session_selector import make_session_selector

# Dashboard parameters: keep these aligned with the current event schema and one-step workflow.
FRONT_SUSPENSION_SELECTOR = {"end": "front", "domain": "wheel"}
REAR_SUSPENSION_SELECTOR = {"end": "rear", "domain": "wheel"}
FRONT_EVENT_SIGNAL_SELECTOR = {"end": "front", "domain": "wheel"}
REAR_EVENT_SIGNAL_SELECTOR = {"end": "rear", "domain": "wheel"}

SCATTER_COMPRESSION_EVENT_ID = "compressions_all>25"
SCATTER_REBOUND_EVENT_ID = "rebounds_all>25"
SCATTER_X_METRIC = "m_peak_disp_max"
SCATTER_COMPRESSION_Y_METRIC = "m_interval_vel_max"
SCATTER_REBOUND_Y_METRIC = "m_interval_vel_min"

runtime_settings_editor = make_preprocess_runtime_settings_editor(
    artifacts_dir=Path("artifacts"),
    bike_profile_path=None,
    fit_dir=None,
    fit_bindings_path=None,
    prompt_for_descriptions=False,
    show_preprocess_profile_path=False,
    show_generic_log_metadata=False,
    show_logger_timezone=False,
    show_bike_profile_path=False,
    show_fit_inputs=False,
    show_prompt_for_descriptions=False,
    show_run_tz_label=False,
)
display(runtime_settings_editor.ui)
```

```python
runtime_settings = runtime_settings_editor.get_settings()
ARTIFACTS_DIR = runtime_settings["artifacts_dir"]

sel = make_session_selector(artifacts_dir=ARTIFACTS_DIR, select_first_by_default=True)
display(sel["ui"])

dashboard = make_simple_suspension_metrics_dashboard(
    sel,
    front_displacement_selector=FRONT_SUSPENSION_SELECTOR,
    rear_displacement_selector=REAR_SUSPENSION_SELECTOR,
    front_velocity_selector=FRONT_SUSPENSION_SELECTOR,
    rear_velocity_selector=REAR_SUSPENSION_SELECTOR,
    front_event_signal_selector=FRONT_EVENT_SIGNAL_SELECTOR,
    rear_event_signal_selector=REAR_EVENT_SIGNAL_SELECTOR,
    compression_event_type=SCATTER_COMPRESSION_EVENT_ID,
    rebound_event_type=SCATTER_REBOUND_EVENT_ID,
    scatter_x_metric=SCATTER_X_METRIC,
    compression_y_metric=SCATTER_COMPRESSION_Y_METRIC,
    rebound_y_metric=SCATTER_REBOUND_Y_METRIC,
)
display(dashboard["ui"])
```
